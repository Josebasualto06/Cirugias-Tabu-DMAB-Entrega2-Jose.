# Comparación de tiempo: CPU local y Colab

Prueba de rendimiento de Tabu uniforme (TS-U) y Tabu con D-MAB completo (TS-DMAB). Se ejecutan 120 corridas: 12 instancias, semillas 2100–2104 y dos métodos, con 6.060 evaluaciones por corrida. Un solo proceso; calentamiento excluido.

Para reproducir: subir `oii464-smoke.zip` al almacenamiento de la sesión y ejecutar las tres celdas en orden. La primera verifica la huella e instala NumPy 2.3.5 y PyYAML 6.0.3; la segunda realiza el benchmark; la tercera descarga sus registros. Usar CPU estándar sin GPU. Las versiones de Python y el hardware se registran en el manifiesto.

El tiempo comparable suma las llamadas completas de búsqueda, incluida su validación final, y excluye instalación, carga de datos, calentamiento y descarga. Los resultados de calidad se guardan para comprobar reproducibilidad entre entornos; este bloque no sustituye la campaña de evaluación del informe.

```python
import sys, subprocess, zipfile, hashlib
from pathlib import Path
package = Path('/content/oii464-smoke.zip')
assert hashlib.sha256(package.read_bytes()).hexdigest() == '8b17719663e3a24ae015459e4bdb11057e9f952384556da0212a8a4f069b46ed'
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'numpy==2.3.5', 'PyYAML==6.0.3'], check=True)
with zipfile.ZipFile(package) as bundle:
    bundle.extractall('/content')
print('Paquete verificado. Python:', sys.version)
print('Benchmark listo: 120 corridas, CPU, un proceso.')
```
